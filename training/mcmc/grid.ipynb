```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Scenario: You drilled 9 boreholes. 6 hit Class III rock.
# Question: What is the TRUE probability of Class III?

p_grid = np.linspace(0,1, 100)
# between 0 and 1, i need 100 samples equally spaced
# grid of 100 candidate probabilities
# its like an array of possible answers of true probab

prior = np.ones(100)
# array of all ones, all probab are equally likely
# initially we are saying, i believe all of the answers above
#  to assign equal weight to every candiate probab

likelihood = binom.pmf(k=6, n=9, p=p_grid)
# in 9 trials, 6 success, probab mass
# p = our hypothesis, wil check of each p_grid[i] as our possible answer
# how likely is our thesis

unstdd_posterior = likelihood*prior
posterior = unstdd_posterior / np.sum(unstdd_posterior)
# need to normalize to 1

display(p_grid)
display(prior)
display(likelihood)
display(posterior)


fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(p_grid, prior, 'k-')
axes[0].set_title("Prior (Uniform)")
axes[0].set_xlabel("P(Class III)")

axes[1].plot(p_grid, likelihood, 'b-')
axes[1].set_title("Likelihood (6/9 hits)")
axes[1].set_xlabel("P(Class III)")

axes[2].plot(p_grid, posterior, 'r-')
axes[2].set_title("POSTERIOR = Prior × Likelihood")
axes[2].set_xlabel("P(Class III)")
axes[2].axvline(6/9, color='gray', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

```

```python
#pessmistic prior

prior_pess = np.where(p_grid<0.5, 2, 0.5 )
# this is not yet a probab, just weight distri, nnormalized to probab via division below
posterior_pess = (likelihood * prior_pess)
posterior_pess = posterior_pess / np.sum(posterior_pess)

plt.plot(p_grid, posterior, 'r-', label='Uniform Prior')
plt.plot(p_grid, posterior_pess, 'b--', label='Pessimistic Prior')
plt.legend()
plt.title("How the Prior Shifts the Posterior")
plt.show()
```
