```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
np.random.seed(42)

# before we use PyMC to find the hidden parameters (μ,σ) from the data (Inverse Problem), we must use NumPy to generate data from known parameters (Forward Problem).

# 1. The Hidden Truth (Only god knows this)
TRUE_MU = 45.0
TRUE_SIGMA = 8.0

# engineers drill 20 boreholes

bore_data = np.random.normal(loc=TRUE_MU, scale=TRUE_SIGMA, size=20)
print(np.mean(bore_data))

# which is NOT exactly 45 and called sampling error, tho it tends to 45
# once our borehole counts increase to like 200K which is not feasible

bore_data_3 = np.random.normal(loc=TRUE_MU, scale=TRUE_SIGMA, size=3)
print(np.mean(bore_data_3))
# wildly disconnected

bore_data = np.append(bore_data, [12])
# massive shear zone, what happens to the mean?
print(np.mean(bore_data))

# sensitive to outliers, mean drops significantly
# students t distro imminent



# Plotting the Forward Problem
plt.figure(figsize=(10, 6))

# Theoretical "True" Distribution Curve
x = np.linspace(TRUE_MU - 4*TRUE_SIGMA, TRUE_MU + 4*TRUE_SIGMA, 200)
pdf = stats.norm.pdf(x, TRUE_MU, TRUE_SIGMA)
plt.plot(x, pdf, label="True Distribution", color='red', lw=2)

# Rug Plot: Data points as markers on the x-axis
plt.plot(bore_data, np.zeros_like(bore_data), 'k|', ms=15, label="Borehole Data (N=20)")

plt.title("Forward Problem: Hidden Truth vs. Ground Samples")
plt.xlabel("MPa (Strength)")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.3)
plt.show()


```
