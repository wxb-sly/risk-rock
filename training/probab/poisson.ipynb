```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

np.random.seed(42)

#lamda = expected rate

# avg of 2 major rockburst per km (where is the per km part?)
rpk = np.random.poisson(lam=4.2, size = 10000)
# output is always an integer

# how to discretize the bins properly
# bins must map 1:1 to integers, +2 ensures that final integer has a complete 1 unit container or else will truncate one int before
# -5 to shift bins to center

bins = np.arange(0, rpk.max() + 2) - 0.5
plt.hist(rpk, bins=bins, density=True, alpha=0.7, color='r', edgecolor='black')
plt.title("Poisson Distribution: Rockbursts per km ($lambda=2$)")
plt.xlabel("Number of Rockbursts")
plt.ylabel("Probability")
plt.xticks(range(0, rpk.max()+1))
plt.show()

probab8 = (rpk>=8).mean()


# Probability of exactly 4 bursts vs 5 bursts when avg is 4.2
p_4 = poisson.pmf(4, mu=4.2)
p_5 = poisson.pmf(5, mu=4.2)

print(f"P(X=4): {p_4:.4f}") # Peak probability is often near the decimal lambda
print(f"P(X=5): {p_5:.4f}")

print(probab8*100) # tail rsik
```

```python
# lamda scales linearly with the domain
# to model distro for 5km drive when avg for 1km is 2 is
# 2 * 5

drive = np.random.poisson(lam= 10, size = 1000000)

bins = np.arange(0, drive.max() + 2) - 0.5
plt.hist(drive, bins=bins, density=True, alpha=0.7, color='r', edgecolor='black')
plt.title("Poisson Distribution: Rockbursts Distro for 5km ($lambda=10$)")
plt.xlabel("Number of Rockbursts")
plt.ylabel("Probability")
plt.xticks(range(0, drive.max()+1, 2))
# add steps to decluuter
plt.tight_layout()
plt.show()
```
